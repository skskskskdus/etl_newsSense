```python
#HTML 태그, 특수문자 제거(bbc 데이터)
import re

def clean(text):
    """
    텍스트(문자열)에서 HTML 태그와 특수문자를 제거하는 함수, 중복된 공백을 단일 공백으로 정리한 뒤 반환 
    """
    # 1) HTML 태그 제거
    #    <...> 형태의 모든 문자열을 공백으로 치환
    text = re.sub(r'<[^>]+>', '', text)

    # 2) 불필요한 특수문자 제거
    #    여기서는 괄호, 특수기호, 이모티콘, 광고용 기호 등을 예시로 처리했습니다.
    #    필요에 따라 범위를 조정하세요.
    text = re.sub(r'[\(\)\[\]\{\}<>]', '', text)
    text = re.sub(r'[#@*■…▶▲△☆★]', '', text)

    # (선택) 알파벳, 숫자, 한글, 일반적인 문장부호 외의 모든 문자 제거
    #text = re.sub(r'[^a-zA-Z0-9가-힣\s.,!?]', '', text)

    # 3) 여러 개의 공백을 하나로 줄이기
    text = re.sub(r'\s+', ' ', text)

    # 앞뒤 공백 제거 후 반환
    return text.strip()

# df.select_dtypes(include='object')는 문자열(오브젝트) 타입 컬럼들만 골라냄
text_cols = df_bbc.select_dtypes(include='object').columns

for col in text_cols:
    df_bbc[col] = df_bbc[col].apply(clean)

df_bbc.head()
print(df_bbc.head)
```

```python
#HTML 태그, 특수문자 제거(sbs 데이터)
# df.select_dtypes(include='object')는 문자열(오브젝트) 타입 컬럼들만 골라냄
text_cols = df_bbc.select_dtypes(include='object').columns

for col in text_cols:
    df_sbs[col] = df_sbs[col].apply(clean)

df_sbs.head()
print(df_sbs.head)
```

```python
#전처리 함수 정의
from nltk.corpus import stopwords as nltk_stopwords
import string

stopwords = set(nltk_stopwords.words('english'))
punctuations = set(string.punctuation)

def preprocess_english(text):
    if not isinstance(text, str):
        return []
    
    # 1.소문자 변환
    text = text.lower()
    # 2.토큰화
    #tokens = word_tokenize(text)
    tokens = text.split()
    # 3. 불용어, 문장 부호 제거
    tokens = [token for token in tokens if token not in stopwords and token not in punctuations]

    return tokens

# bbc 뉴스 데이터 적용: Content 열에 대해 전처리 수행
df_bbc['Content_tokens'] = df_bbc['Content'].apply(preprocess_english)

#결과 확인
print(df_bbc[['Content', 'Content_tokens']].head(5))
```

# 형태소 분석,불용어 제거(sbs 데이터)

```python
# 1. 라이브러리 설치 (최초 한 번만 실행)
#! pip install konlpy
```

```python
# 2. 라이브러리 불러오기
from konlpy.tag import Okt

# 3. 분석기 생성
okt = Okt()

# 4. 형태소 분석 함수 정의
def tokenize_text(text):
    if not isinstance(text, str):  # NaN 또는 숫자 예외 처리
        return []
    return okt.morphs(text, stem=True)  # stem=True: 어간 추출 (예: '합니다' → '하다')
```

```python
# 5. 전체 데이터에 적용 (예: content 컬럼에 형태소 분석 적용)
df_bbc['tokens'] = df_bbc['Content'].apply(tokenize_text)
```

```python
# 6. 불용어 사전 정의
stopwords = [
    '은', '는', '이', '가', '을', '를', '에', '의', '와', '과', '으로', '해서', '으로서',
    '하다', '되다', '있다', '없다', '이다', '것', '수', '등', '더', '같다',
    '그리고', '그러나', '또는', '하지만', '고', '도', '또', '및', '중',
    '대한', '한', '해', '를', '되었', '했다'
    # 필요시 계속 추가 가능!
]

# 7. 불용어 제거 함수 정의
def remove_stopwords(tokens):
    return [t for t in tokens if t not in stopwords]

# 8. 전체 데이터에 적용
df_bbc['tokens_no_stop'] = df_bbc['tokens'].apply(remove_stopwords)
```
